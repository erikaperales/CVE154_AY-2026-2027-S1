# Programming Activity 01. Implementing mathematical functions

*Last updated 04 September 2026*

## Scenario

A watering trough of length $l$ has a semicircular cross section of diameter $d$.
You are to determine:

- the [wetted perimeter](https://en.wikipedia.org/wiki/Wetted_perimeter) $P$
  and
- the water volume $V$ in the trough

whenever the top of the trough is a distance $h$
above the water surface.
The trough is not sloped.

Mathematically,
you need to express $P$ and $V$ as functions of $l$, $d$, and $h$.

## Task

Define three functions `watvol`, `wetper`, and `volper`.

`watvol` has three required (positional) arguments:

1. the radius of the cross section,
2. the distance $h$,
   and
3. the trough length,

and outputs the water volume.
Take the input dimensions to be of the same units.

`wetper` has two required (positional) arguments:

1. the diameter of the cross section,
   and
2. the distance $h$,

and outputs the wetted perimeter.
Again, assume dimensional homogeneity is maintained.

`volper` has two required (positional) arguments:

1. the diameter (in inches) of the cross section,
2. the trough length (in meters),
   and
3. the distance $h$ (in centimeters),

and outputs a list of two elements:

1. the water volume (in cubic meters)
   and
2. the wetted perimeter (in inches).

Observe the following special requirements.

1. Do not import any module or package other than Python's [`math`](https://docs.python.org/3/library/math.html).
2. Use `watvol` and `wetper` in defining `volper`.

## Marking

Correct implementations of `watvol` and of `wetper` each earns 5 points.
Correct implementation of `volper` earns 10 points.

The total mark is computed as
$0.9^{a} 0.8^{b} r$,
where:

- $r$ is the raw score,
- $b$ is the number of special requirements not met,
  and
- $a$ is the number of general requirements not met.
