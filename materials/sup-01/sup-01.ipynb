## On preparing the submission script

The template script will have three main parts.

1. The *docstring* take up the first few lines,
   starting with `"""`
   (*i.e.*, three double quotes)
   and ending with the same.
2. The *synthesis* take up the last lines of the script,
   starting with the line `if __name___ == "__main__":`.
3. The *core* is that sandwiched by the docstring and the synthesis.

Apart from the docstring,
these names are for use in this course only.

The following are some common rules for preparing the submission script for any activity.
Each violation means a demerit.

Do not modify the docstring.

When the activity does not require you to write code in the synthesis,
leave it as it is given in the template script.

When tasked to write code in the synthesis,
do so *after* the line
`if __name___ == "__main__":`,
observing [Python's indentation rules](https://realpython.com/how-to-indent-in-python/).

Much, if not all, of the code you will write will be the core.
(As such, you can think of the template script as having an "empty" core.)

All [imports](https://realpython.com/ref/best-practices/imports/)
must occupy the top few lines of the core.

Directly following the imports is a line
declaring a string that spells out your name
(given and surname *only*),
stored to a [variable](https://realpython.com/python-variables/) `__AUTHOR__`.
For instance, for a student named Rain L. Niño,
the variable assignment would be
`__AUTHOR__ = "Rain Niño"`.
